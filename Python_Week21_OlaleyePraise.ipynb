```python

balance = 0


def deposit():
    """This machine puts money INTO the bank."""
    global balance 
    amount = float(input("How much do you want to deposit? # 4"
                        "Account Number to deposit - 3489072869"))

    if amount <= 0:
        print("❌ You can't deposit 0 or negative money")
    else:
        balance = balance + amount  
        print(f"✅ You deposited #{amount:.2f}. New balance: #{balance:.2f}")

def send ():
    """This machine sends money to another bank."""
    global balance
    
    amount = float(input("How much do you want to send? #"))

    if amount <= 0:
        print("❌ You can't send 0 or negative money")
 
    elif amount > balance:
        print(f"❌ You only have #{balance:.2f}. You can't take out that much!")
    else:
        Account_number = input(" Enter the 10 digits account number:")
        
        while len(Account_number)!= 10 or not Account_number.isdigit():
          print("❌ Account Number must be exactly 10 digits. Try again.")
          Account_number = input(" Enter the 10 digits account number:")
            
        balance = balance - amount 
        print(f"✅ Successful!")
        print(f" Amount: #{amount:.2f}.")
        print(f" Account Number: {Account_number}")
        print(f" New balance: #{balance:.2f}")


def check_balance():
    """This machine just tells you what's in the box."""
    print(f"💰 Your current balance is: #{balance:.2f}")



def main():
    print("=======================================")
    print(" WELCOME TO PRAISE MICROFINANCE BANK 🏦")
    print("=======================================")

    still_running = True  
    while still_running:
        print("\nWhat would you like to do?")
        print("1. Deposit money")
        print("2. Send money")
        print("3. Check balance")
        print("4. Quit")

        choice = input("Type a number (1-4): ")

        if choice == "1":
            deposit()
        elif choice == "2":
            send()
        elif choice == "3":
            check_balance()
        elif choice == "4":
            print("👋 Thanks for banking with us. Bye!")
            still_running = False 
        else:
            print("❌ I didn't understand that. Please type 1, 2, 3, or 4.")

if __name__ == "__main__":
    main()
```
